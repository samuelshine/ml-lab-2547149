# Distance Matrix Computation

Demonstrating distance matrix computation using Euclidean, Manhattan, and Minkowski distances.

```
import pandas as pd
from sklearn.metrics import pairwise_distances

# Load dataset
df = pd.read_excel('data/lab2_data.xlsx')
print(df.head())
```

```
# Select numeric columns for distance computation
numeric_df = df.select_dtypes(include=['number'])

# If there are NaN values, we should fill or drop them
numeric_df = numeric_df.dropna()

print("Shape of numeric data used for distance computation:", numeric_df.shape)
```

```
# 1. Euclidean Distance
euclidean_dist = pairwise_distances(numeric_df, metric='euclidean')
print("Euclidean Distance Matrix:
", euclidean_dist)
```

```
# 2. Manhattan Distance
manhattan_dist = pairwise_distances(numeric_df, metric='manhattan')
print("Manhattan Distance Matrix:
", manhattan_dist)
```

```
# 3. Minkowski Distance
minkowski_dist = pairwise_distances(numeric_df, metric='minkowski', p=3)
print("Minkowski Distance Matrix (p=3):
", minkowski_dist)
```
