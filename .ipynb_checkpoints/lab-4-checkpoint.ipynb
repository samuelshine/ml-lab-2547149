# Lab 4: KNN Classification & Evaluation Metrics

## Problem Statement
Build a KNN classifier for early cancer detection, optimize K, and compare classification metrics with regression metrics.

## Task 1: Data Preparation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_curve, auc

df = pd.read_csv('data/brca.csv')
if df.columns[0] == 'Unnamed: 0' or df.columns[0] == '':
    df = df.iloc[:, 1:]

print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

# M=0 (Malignant), B=1 (Benign)
df['y'] = df['y'].map({'M': 0, 'B': 1})

X = df.drop('y', axis=1)
y = df['y']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

**Justification for Feature Scaling:**
KNN relies on distance. Scaling prevents features with large numeric ranges from dominating distance calculations.

## Task 2: Train-Test Split Analysis

```python
splits = [0.2, 0.3, 0.1]
for test_size in splits:
    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=test_size, random_state=42, stratify=y)
    knn = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
    print(f"Split {int((1-test_size)*100)}:{int(test_size*100)} -> Train Acc: {knn.score(X_train, y_train):.4f}, Test Acc: {knn.score(X_test, y_test):.4f}")
```

**Analysis:**
Different splits cause accuracy variance. Larger training sets improve learning but leave fewer samples for reliable testing.

## Task 3: KNN Model & Heuristic K

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)
n = X_train.shape[0]
hk = int(np.sqrt(n))
hk = hk if hk % 2 != 0 else hk + 1
print(f"Heuristic K (√n): {hk}")

k_vals = list(range(max(1, hk - 10), hk + 11, 2))
train_acc, test_acc = [], []

for k in k_vals:
    knn = KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train)
    train_acc.append(knn.score(X_train, y_train))
    test_acc.append(knn.score(X_test, y_test))

plt.plot(k_vals, train_acc, marker='o', label='Train')
plt.plot(k_vals, test_acc, marker='s', label='Test')
plt.axvline(hk, color='r', linestyle='--', label=f'Heuristic K={hk}')
plt.legend(); plt.grid(); plt.show()
```

**Distance Metrics:**
- **Euclidean:** Straight-line distance.
- **Manhattan:** Sum of absolute differences (grid-based).

**Decision Boundaries:**
- **Small K:** Complex, jagged boundaries (Overfitting).
- **Large K:** Smoother boundaries, eventually predicting the majority class (Underfitting).

## Task 4: Cross Validation

```python
cv_scores = [cross_val_score(KNeighborsClassifier(n_neighbors=k), X_scaled, y, cv=5).mean() for k in k_vals]

plt.plot(k_vals, cv_scores, marker='d', color='g', label='5-Fold CV')
plt.axvline(hk, color='r', linestyle='--', label=f'Heuristic K={hk}')
plt.legend(); plt.grid(); plt.show()

best_k = k_vals[np.argmax(cv_scores)]
print(f"Best K from CV: {best_k}")
```

## Task 5: Classification Evaluation

```python
knn_final = KNeighborsClassifier(n_neighbors=best_k).fit(X_train, y_train)
y_pred = knn_final.predict(X_test)
y_prob = knn_final.predict_proba(X_test)[:, 1]

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred, pos_label=0):.4f}")
print(f"Recall: {recall_score(y_test, y_pred, pos_label=0):.4f}")
print(f"F1 Score: {f1_score(y_test, y_pred, pos_label=0):.4f}")

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix'); plt.show()

fpr, tpr, _ = roc_curve(y_test, y_prob, pos_label=1)
plt.plot(fpr, tpr, label=f'AUC = {auc(fpr, tpr):.2f}')
plt.plot([0, 1], [0, 1], 'k--'); plt.legend(); plt.show()
```

## Task 6: Comparative Study

- **Error (Regression) vs Decision (Classification):** Regression measures error magnitude (distance from truth). Classification measures prediction correctness (discrete categories).
- **Comparisons:**
  - **R² vs Accuracy:** R² = Variance explained; Accuracy = % of correct classes.
  - **RMSE vs F1 Score:** RMSE = Penalizes large numerical errors; F1 = Balances false positives/negatives.
  - **MAE vs Confusion Matrix:** MAE = Average absolute error; Confusion Matrix = Exact breakdown of misclassifications.

**Inference:**
- **Accuracy Shortfall:** In medical data, classes are imbalanced. High accuracy can still mean missing all cancer cases.
- **Recall & ROC-AUC:** Recall minimizes missed cancer diagnoses (False Negatives). ROC-AUC evaluates class separation across all thresholds.

## Task 7: Analytical Questions
- **Lazy Learner:** Memorizes training data instead of building a predefined model.
- **Feature Scaling:** Ensures large-scale features don't dominate distance calculations.
- **√n Rule:** Provides a balanced baseline K that scales with data size.
- **Cross-Validation:** Reduces variance by testing on multiple data splits.
- **Bias-Variance:** Small K = High Variance (Overfit); Large K = High Bias (Underfit).
- **Recall Importance:** Missing a cancer diagnosis is worse than a false alarm.
- **Large K Limit:** Defaults to predicting only the majority class.

## Conclusion
- **Optimal K:** Found using heuristic (√n) and confirmed via CV (typically ~15-21).
- **Splits:** Variations proved the necessity of cross-validation.
- **Metrics:** Recall and ROC-AUC are far more critical than accuracy in healthcare classification tasks.
